## This Jupyter Notebook is designed to demonstrate the Pachner Moves functions with visualization

Let's start with the first one and check how the 2-2 move works, in the first image, you'll see a regular initial piece; in the next one, you'll see the same piece but with a modified diagonal
Also, it's worth keeping in mind that the piece can change depending on which vertices the algorithm chooses for the diagonal
After images user can see the prints that shows the difference between two triangles

```python
import PySimplicial.utils as ps

octahedron = [(0,1,2), (0,2,3), (0,3,4), (0,4,1),(5,2,1), (5,3,2), (5,4,3), (5,1,4)] #8 triangles list
ps.visualize_triangulation_2D(octahedron)

octahedron_2_2 = ps.move_2_2(octahedron)
ps.visualize_triangulation_2D(octahedron_2_2)

print(f"Octahedron={octahedron}")
print(f"Octahedron with move 2-2={octahedron_2_2}")
```

Now we need to introduce the following type of triangulation:

**Move 1-3 and its inverse, 3-1**

That is, we add *one new internal vertex* (the boundaries remain the same) and connect it to the other three, as if *dividing the triangle into three smaller ones*. Then, we find the point and remove it and all the connections it made, *as if returning the original triangle.*

Just like with the previous version, you will see prints below that will show that at the end we will return to the original figure

```python
# octahedron is [(0,1,2), (0,2,3), (0,3,4), (0,4,1),(5,2,1), (5,3,2), (5,4,3), (5,1,4)]
ps.visualize_triangulation_2D(octahedron)

Pachner_move_1_3 = ps.move_1_3(octahedron)
ps.visualize_triangulation_2D(Pachner_move_1_3)

Pachner_move_3_1 = ps.move_3_1(Pachner_move_1_3)
ps.visualize_triangulation_2D(Pachner_move_3_1)


print(f"Basic octahedron={octahedron}")
print(f"Octahedron with move 1-3={Pachner_move_1_3}")
print(f"Octahedron with move 3-1={Pachner_move_3_1}")
```

Now, let's move on to **three-dimensional space**. Namely, to **tetrahedron triangulations**

There are **four types**, two of which are inverse

*"1-4"* - divides one tetrahedron into four smaller tetrahedrons by introducing a new internal vertex

*"4-1"* - the inverse of 1-4 and removes an internal vertex that is a common vertex of exactly four tetrahedra with no other internal simplices, collapsing them back into a single tetrahedron

*"2-3"* - increases the number of tetrahedra by one and adds one new edge

*"3-2"* - the inverse of the 2-3 move, replacing three tetrahedra meeting on an internal edge between two tetrahedra sharing a face, reducing the number of tetrahedra by one, and removing one edge without changing vertices

Now, a visualization of "1-4" and its inverse "4-1" will be **shown**

As usual, the prints will show the **difference** between the two triangles

```python
one_tetrahedron = [(0, 1, 2, 3)]
ps.visualize_triangulation_3D(one_tetrahedron)

Pachner_move_1_4 = ps.move_1_4(one_tetrahedron)
ps.visualize_triangulation_3D(Pachner_move_1_4)

Pachner_move_4_1 = ps.move_4_1(Pachner_move_1_4)
ps.visualize_triangulation_3D(Pachner_move_4_1)

print(f"one tetrahedron={one_tetrahedron}")
print(f"tetrahedron with move 1-4={Pachner_move_1_4}")
print(f"Pachner move 4-1 on tetrahedron with move 1-4={Pachner_move_4_1}")
```

Now, a visualization of "2-3" and its inverse "3-2" will be **shown**

As usual, the prints will show the **difference** between the two triangles

```python
tetrahedron = [(0, 1, 2, 3),(0, 1, 2, 4)] # Visualize original figure
ps.visualize_triangulation_3D(tetrahedron)

Pachner_move_2_3 = ps.move_2_3(tetrahedron) # visualize the same figure but with Pachner Move 2-3 triangulation
ps.visualize_triangulation_3D(Pachner_move_2_3)

Pachner_move_3_2 = ps.move_3_2(Pachner_move_2_3) # return this modify to basic form
ps.visualize_triangulation_3D(Pachner_move_3_2)

print(f"tetrahedron={tetrahedron}")
print(f"same tetrahedron but with move 2-3={Pachner_move_2_3}")
print(f"inverse, move 3-2={Pachner_move_3_2}")
```
